# **SCATTER PLOT**

```python
sns.scatterplot(x="Daily_Social_Media_Hours",
                y = "Mental_Health_Score",
                hue = "Gender",
                data = df )
plt.title("social media vs mental health")
plt.savefig("scatterplot.png", bbox_inches = "tight",dpi = 300)
plt.show()
```

# **INTERPRETATION**
* students spend more hours on social media every day, their mental health
scores tend to drop.

* ​Students who keep their social media time low—between 0 to 4
hours—consistently show higher and healthier well-being scores.

* ​Once screen time crosses 8 hours or more, mental health scores noticeably slide down into the lower, more distressed range.

* ​This negative trend affects male, female, and non-binary students pretty equally, meaning high screen time is a universal risk factor across all groups.

# **JOINT PLOT**

```python
sns.jointplot( x= "Daily_AI_Tool_Usage_Hours",
              y= "Mental_Health_Score",
              data = df )
plt.savefig("jointplot.png", bbox_inches = "tight")
plt.show()
```

# **INTERPRETATION**
* The bar diagram on the right shows that most students have mental health scores clustered heavily in the middle-to-high range (around 70 to 80).

* ​Very few students drop down to the lower mental health score zone (near 50), which means severe psychological distress is less common in this dataset.

* ​The bar diagram at the top shows how many students fall under each specific hour category on the X-axis.

* ​Putting it all together, these side bars let you see the overall population count on the edges while checking how those scores are distributed in the middle scatter plot.

# **BOX PLOT**

```python
sns.boxplot(x = "Gender",
            y = "Physical_Activity_Hours",
            data = df,
            palette="Set2")
plt.savefig("boxplot.png", bbox_inches = "tight")
plt.show()
```

# **INTERPRETATION**
* This box plot compares Physical Activity Hours across the three Gender categories (Male, Non-binary, and Female).

* ​The middle line (median) sits right at 1.0 hour for all three groups, showing that typical physical activity levels are identical across genders.

* ​The boxes and whiskers stretch from 0 up to nearly 5 hours, meaning the overall spread of activity habits is very similar regardless of gender.

* ​This tells you there is no major gender-based difference in daily physical activity among the students in your dataset

```python
sns.scatterplot(x="Physical_Activity_Hours",
                y = "Physical_Health_Score",
                hue = "Education_Level",
                data = df )
plt.savefig("scatterplot2.png", bbox_inches = "tight")
plt.show()
```

# **INTERPRETATION**
* As you look from left to right (from 0 to 5 hours of physical activity), the lowest dots move higher up. This means students who exercise more tend to have higher physical health scores.

*    The blue, red, and green dots are mixed together in every single column. This shows that High School, College, and University students all share a very similar pattern when it comes to exercise and physical health.



# **HEAT MAP**

```python
select_col = ["Age",
              "Daily_Social_Media_Hours",
              "Daily_AI_Tool_Usage_Hours",
              "Sleep_Hours",
              "Physical_Activity_Hours",
              "Mental_Health_Score",
              "Physical_Health_Score"]
cor_col = df[select_col].corr()

sns.heatmap(cor_col,
            annot= True, # show the exact number inside the box
            cmap = "coolwarm", #red for positive, blue for negative
            fmt= ".2f" # fix decimal points
            )
plt.savefig("heatmap.png", bbox_inches = "tight")
plt.show()
```

# **INTERPRETATION**
* Screen Time Impact (-0.40 and -0.31): Daily social media hours show negative correlations with both Mental Health Score (-0.40) and Physical Health Score (-0.31). This is your strongest negative pattern.

* ​The Exercise Benefit (+0.41): Physical activity hours show a strong positive link with Physical Health Score (+0.41), proving that active routines directly support better physical well-being.

* ​The Sleep Connection (+0.30 and -0.29): Sleep hours positively affect mental health (+0.30) while social media negatively affects sleep (-0.29), showing a chain reaction where excessive screen time cuts into sleep, which in turn lowers mental health.

* ​Age Independence (0.00): Age has virtually no correlation with habits or health scores, meaning these trends apply evenly across students regardless of how old they are.

# **VIOLIN PLOT**

```python
sns.violinplot(
    data=df,
    x="Education_Level",
    y="Daily_Social_Media_Hours",
    palette="muted",
    inner="box", # it put box inside the violin
)
plt.savefig("violinplot.png", bbox_inches = "tight")
plt.show()
```

# **INTERPRETATION**

The violin plot reveals that daily social media usage patterns are nearly identical across all education levels (College, High School, and University). Each group exhibits a median daily screen time of approximately 4 to 5 hours, with the vast majority of students falling between 3 and 6 hours per day. Education level does not significantly impact screen time duration among students in this dataset.

# **STRIP PLOT**

```python
sns.stripplot(
    data=df,
    x="Gender",
    y="Daily_AI_Tool_Usage_Hours",
    hue="Gender",
    palette="Set2",
    jitter=0.2, #instead of showing messy dots it creates the horizontal line
    size=6,
    alpha=0.7, # dot's density
)
plt.savefig("striplot.png", bbox_inches = "tight")
plt.show()
```

# **INTERPRETATION**
The strip plot demonstrates that daily AI tool usage hours are evenly distributed across all gender groups (Male, Non-binary, and Female). Usage spans consistently from 0 to nearly 10 hours per day, indicating that gender identity does not create a barrier or bias toward adopting AI tools.
